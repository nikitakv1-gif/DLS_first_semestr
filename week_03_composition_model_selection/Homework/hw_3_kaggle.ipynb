```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
```

```python
X = data.drop(columns = 'Churn')
y = data['Churn']

X_train, X_val, y_train, y_val = train_test_split(X, y, shuffle = True, stratify = y, test_size = 0.2)
```

```python
#ураааа, учим модель

#содаем пайплайн, коллеги
preprocess = ColumnTransformer(transformers = [('standard', StandardScaler(), num_cols_e),
                                               ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False), cat_cols)])

pipeline = make_pipeline(preprocess, LogisticRegression())

grid = GridSearchCV(pipeline,
                    param_grid = {'logisticregression__C': [100, 10, 1, 0.1, 0.01, 0.001]},
                    cv = 5,
                    n_jobs=-1,
                    scoring='roc_auc')

fit_res = grid.fit(X_train, y_train)
```

- Лучшие параметры {'logisticregression__C': 10}
